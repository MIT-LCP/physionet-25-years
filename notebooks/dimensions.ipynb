## Querying the API

````python
import requests

# Define your Dimensions API key
api_key = "YOUR_DIMENSIONS_API_KEY"

# Define the query
query = 'search full_data for "PhysioNet" OR "MIMIC-III"'

# Define the API endpoint
url = "https://app.dimensions.ai/api/dsl.json"

# Set the headers
headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
}

# Send the request
response = requests.post(url, headers=headers, json={"query": query})
data = response.json()

# Example output: number of matching documents
print(f"Total documents found: {len(data['documents'])}")

# Print out the titles of the first few documents
for doc in data['documents'][:5]:
    print(doc['title'])
````
