````python
# construct the query
search_terms = (
    "physionet",
    "MIMIC-III",
    "MIMIC-IV",
    "PhysioNet",
    "mimic-iv",
    "Physionet",
    "MIT-BIH",
    "WFDB",
    "10.13026",
    "MIMIC-CXR",
    "physiotoolkit",
    "physiobank",
    "eICU-CRD"
)

# Join the terms with " OR "
search_query = " OR ".join(search_terms)

query = (
        f'search publications '
        f'in full_data_exact for "{search_query}" '
        f'return publications[basics]'
)

print(query)
````

````python
# get result
result = dsl.query(query, verbose=False)
````

````python
# print some stats using the Result object
print("Results in this batch: ", result.count_batch)
print("Results in total: ", result.count_total)
print("Errors: ",result.errors)
````
