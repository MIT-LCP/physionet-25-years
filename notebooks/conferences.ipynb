# Searching for keywords in conference proceedings

This notebook looks at how PhysioNet resources are being used in conference proceedings.

## Setup

```python
import os
import pandas as pd
from twentyfiveyears.models import (PublicationVenue,
                                    Publication,
                                    VenueCollection)
```

## Create list of conferences

We'll first create the following objects:

- List of conferences (`PublicationVenue`)
- List of papers associated with each conference (`Publication`)
- A collection of conferences (`VenueCollection`)

```python
# make a list of conferences that we would like to explore
conference_list = {"chil": "Conference on Health, Inference, and Learning",
                   "embc": "IEEE Engineering in Medicine and Biology",
                   "iclr": "International Conference on Learning Representations",
                   "icml": "International Conference on Machine Learning",
                   "mlforhc": "Machine Learning for Healthcare",
                   "ml4h": "Machine Learning for Health",
                   "neurips": "Neural Information Processing Systems",
                  }
```

```python
# initialize collection of conferences
conferences = VenueCollection()

# path to papers
base_path = os.path.join("..", "data", "papers", "2023")

# create the collection of conferences
for shortname, title in conference_list.items():
    path = os.path.join(base_path, shortname)
    conference = PublicationVenue(title=title,
                                  shortname=shortname,
                                  year=2023,
                                  path=path)
    conferences.add(conference)
```

## Add publications to conferences

We can now associate a list of publications with each conference.

```python
# for each conference
# add the publications in the conference folder
# unfortunately we aren't able to populate the title, authors, etc.
for conference in conferences:
    for file in os.listdir(conference.path):
        if file.endswith('.pdf'):
            paper = Publication(path=conference.path,
                                filename=file)
            conference.add_publication(paper)
```

```python
print(f"Number of conferences: {len(conferences)}")
```

## Find keywords in publications

Now search for our keywords of interest.

```python
# define a set of keywords to search.
keywords = ["PhysioNet", "physionet"]
```

```python
# search across venues
conferences.search_keywords_across_venues(keywords)
```

## Summarize findings

Summarize the findings of the keyword search.

```python
for conference in conferences:
    conference.summarize()
```
