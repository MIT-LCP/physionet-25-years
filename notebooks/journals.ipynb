# Searching for keywords in journals

This notebook looks at how PhysioNet resources are being used in journals.

## Setup

```python
import os

from twentyfiveyears.models import (PublicationVenue,
                                    Publication,
                                    VenueCollection)
```

## Create list of journals

We'll first create the following objects:

- List of conferences (`PublicationVenue`)
- List of papers associated with each conference (`Publication`)
- A collection of conferences (`VenueCollection`)

```python
# make a list of conferences that we would like to explore
journal_list = {"applied_clinical_health": "applied_clinical_health",
                "bmc_digital_health": "bmc_digital_health",
                "gigascience": "gigascience",
                "jamia": "jamia",
                "jamia_open": "jamia_open",
                "jmir": "jmir",
                "lancet_digital_health": "lancet_digital_health",
                "nature_digital_medicine": "nature_digital_medicine",
                "nature_scientific_data": "nature_scientific_data",
                "plos_digital_health": "plos_digital_health",
               }
```

```python
# initialize collection of conferences
journals = VenueCollection()

# path to papers
base_path = os.path.join("..", "data", "papers", "2023")

# create the collection of conferences
for shortname, title in journal_list.items():
    path = os.path.join(base_path, shortname)
    journal = PublicationVenue(title=title,
                               shortname=shortname,
                               year=2023,
                               path=path)
    journals.add(journal)
```

## Add publications to journals

We can now associate a list of publications with each journal.

```python
# for each journal
# add the publications in the journal folder
# unfortunately we aren't able to populate the title, authors, etc.
for journal in journals:
    for file in os.listdir(journal.path):
        if file.endswith('.pdf'):
            paper = Publication(path=journal.path,
                                filename=file)
            journal.add_publication(paper)
```

```python
print(f"Number of journals: {len(journals)}")
```

## Find keywords in publications¶

Now search for our keywords of interest.

```python
# define a set of keywords to search.
keywords = ["PhysioNet",
            "physionet",
            "Physionet",
            "MIMIC",
            "MIMIC-III",
            "MIMIC-IV"
            "MIMIC-ED",
            "MIMIC-CXR",
            "MIMIC-IV-ED",
            "MIMIC-IV-CXR",
            "MIMIC-IV-ECG",
            "MIMIC-IV-ECHO",
            "wfdb",
            "WFDB",
            "10.13026",
            "eICU-CRD",
            "MIT-BIH",
            "WFDB"
           ]
```

```python
# search across venues
journals.search_keywords_across_venues(keywords)
```

## Summarize findings

Summarize the findings of the keyword search. Print passages around keyword hits if requested.

```python
print_passages = False
for journal in journals:
    journal.summarize()
    if print_passages:
        journal.print_passages()
```
