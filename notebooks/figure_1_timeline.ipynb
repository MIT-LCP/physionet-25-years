# PhysioNet timeline

## Key events:

- PhysioNet was formally established in 1999 as the outreach component of the Research Resource for Complex Physiologic Signals, a cooperative project initiated by a diverse group of computer scientists, physicists, mathematicians, biomedical researchers, clinicians, and educators at Boston's Beth Israel Deaconess Medical Center/Harvard Medical School and MIT
- At the time PhysioNet was established, members of the PhysioNet team at MIT were preparing to host Computers in Cardiology (CinC) 2000. We hoped to introduce PhysioNet to our international colleagues who would be attending CinC
- A timely contribution of data made it possible to create the first PhysioNet/CinC Challenge, which attracted the attention of more than a dozen teams to the subject of detecting sleep apnea from the ECG.
- PhysioNet was originally established under the auspices of the National Center for Research Resources (NCRR) of the National Institutes of Health (NIH). The NCRR was discontinued in 2011 as part of a major NIH reorganization
- National Institute of Biomedical Imaging and Bioengineering (NIBIB) under grant R01EB030362.

```python
def make_timeline(
        events,
        title,
        year_start=1980,
        year_end=2030,
        path_png=None,
        path_svg=None,
        show=False):

    fig, ax = plt.subplots(figsize=(15, 11))

    # Lanes and positions
    lanes = events.keys()
    y_positions = {lane: i for i, lane in enumerate(reversed(lanes))}

    # --- Decade shading (start from first *complete* decade after year_start) ---
    # e.g. year_start=1999 -> first decade shading is 2000s
    decade_min = ((year_start + 9) // 10) * 10
    decade_max = ((year_end // 10) + 1) * 10

    for decade_start in range(decade_min, decade_max, 10):
        color = "#f5f5f5" if (decade_start // 10) % 2 == 0 else "#e9e9e9"
        if decade_start == 2020:
            color = "#fffdd0"

        rect = patches.Rectangle(
            (decade_start, -0.7),
            10,
            len(lanes) + 1.0,
            linewidth=0,
            facecolor=color,
            alpha=0.9,
        )
        ax.add_patch(rect)

    # --- Year micro-ticks ---
    for year in range(year_start, year_end + 1):
        ax.vlines(
            year,
            -0.6,
            len(lanes) - 0.4,
            linewidth=0.3,
            color="#cfcfcf",
            alpha=0.4,
        )

    # --- Lane lines (solid to 2024, dashed to year_end) ---
    PRESENT_YEAR = 2024  # if you want, you can make this a parameter too

    for lane, y in y_positions.items():
        # Solid from start to min(PRESENT_YEAR, year_end)
        ax.hlines(
            y,
            year_start,
            min(PRESENT_YEAR, year_end)+1,
            linewidth=1.5,
            color="#555555"
        )

        # Dashed from PRESENT_YEAR to year_end, if applicable
        if year_end > PRESENT_YEAR:
            ax.hlines(
                y,
                PRESENT_YEAR,
                year_end,
                linewidth=1.5,
                color="#555555",
                linestyles="--",
            )

        # Lane label
        ax.text(
            year_start - 1.0,
            y,
            lane,
            ha="right",
            va="center",
            fontsize=12,
            fontweight="bold",
            color="#222222",
        )

    # --- Events with explicit direction, x-offset, and wrapping ---
    for lane, items in events.items():
        y = y_positions[lane]
        for year, item_title, relevance, direction, x_offset, wrap_width in items:
            # Skip events outside visible year range
            if year < year_start or year > year_end:
                continue

            x = year

            # Vertical displacement (up or down)
            offset_y = 0.28 * direction

            # Horizontal displacement
            text_x = x + x_offset
            ha = "left" if x_offset > 0 else "right"

            # Wrap item_title and relevance text
            title_wrapped = "\n".join(textwrap.wrap(item_title, width=wrap_width))
            relevance_wrapped = "\n".join(textwrap.wrap(relevance, width=wrap_width))

            # Connector line
            ax.vlines(x, y, y + offset_y, linewidth=1.2, color="#555555")

            # Marker
            ax.plot(
                x,
                y,
                marker="o",
                markersize=7,
                markerfacecolor="white",
                markeredgecolor="#555555",
                markeredgewidth=2,
            )

            # Title (bold)
            ax.text(
                text_x,
                y + offset_y + 0.05,
                title_wrapped,
                fontsize=10.2,
                ha=ha,
                va="bottom",
                color="#222222",
                fontweight="bold",
            )

            # Relevance line(s)
            ax.text(
                text_x,
                y + offset_y - 0.03,
                relevance_wrapped,
                fontsize=8.6,
                ha=ha,
                va="top",
                color="#555555",
            )

    # --- Decade labels (also start at first complete decade) ---
    top_y = len(lanes) - 0.1
    for decade_start in range(decade_min, decade_max, 10):
        ax.text(
            decade_start + 5,
            top_y + 0.3,
            f"{decade_start}s",
            ha="center",
            fontsize=11,
            color="#333333",
            alpha=0.8,
        )

    # --- Title ---
    ax.set_title(
        title,
        fontsize=16,
        fontweight="bold",
        pad=0,
    )

    # Layout
    ax.set_ylim(-0.7, len(lanes) + 0.7)
    ax.set_xlim(year_start - 1, year_end)
    ax.axis("off")

    plt.tight_layout()

    # Save outputs
    if path_png:
        plt.savefig(path_png, dpi=300)
    if path_svg:
        plt.savefig(path_svg)

    if show:
        plt.show()
    else:
        plt.close(fig)
```

```python
title="Milestones in the evolution of the PhysioNet platform"
output_path="../figures/figure_1_timeline.png"
make_timeline(events, title, year_start=1995, year_end=2029, path_png=output_path, show=True)
```
